# Edge detection project


## Import libraries

```python
import cv2 as cv    
import matplotlib.pyplot as plt
```

## Add the image

```python
img = cv.imread("Images/little_girl.jpg")

gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)

plt.imshow(gray, "gray")
```

# Sobel


```python
sobelx = cv.Sobel(gray, cv.CV_64F, dx=1, dy=0)
sobely = cv.Sobel(gray, cv.CV_64F, dx=0, dy=1)
mag = cv.magnitude(sobelx, sobely)
sobel = cv.convertScaleAbs(mag)

plt.imshow(sobel, "gray")
```

# Laplacian


```python
blur = cv.GaussianBlur(gray, (3,3), 0)

lap = cv.Laplacian(blur, cv.CV_64F)

lap = cv.convertScaleAbs(lap)

plt.imshow(lap, "gray")
```

# Canny

```python

canny = cv.Canny(gray, threshold1=100, threshold2=200)

plt.imshow(canny, "gray")
```

# visualization

```python

fig, axes = plt.subplots(1,4, figsize=[20, 5])


images = [
    img[..., ::-1],
    sobel,
    lap,
    canny
]

titles = [
    "Normal Image",
    "Sobel Edge Detection",
    "Laplacian Edge Detection",
    "Canny Edge Detection"
]

for ax, image, title in zip(axes, images, titles):
    ax.imshow(image, cmap="gray")
    ax.set_title(title)
    ax.axis("off")

plt.tight_layout()
plt.show()
```
