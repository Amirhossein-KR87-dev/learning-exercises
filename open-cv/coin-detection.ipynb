```python
import numpy as np
import matplotlib.pyplot as plt
import cv2 as cv

# load the image
img = cv.imread("Image/iranian-coin.jpg")
gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
gray = cv.GaussianBlur(gray, (9, 9), 2)
# hough circle
circle = cv.HoughCircles(gray,
    method=cv.HOUGH_GRADIENT,
    dp=1,           
    minDist=120,     
    param1=110,    
    param2=60,    
    minRadius=10,   
    maxRadius=100 )

# draw circles
count = 0
if circle is not None:
    circle = np.uint16(np.around(circle))
    for c in circle[0]:
        x, y, r = c
        count +=1
        cv.circle(img, (x, y), r, (0, 255, 0), 3)

print(f"count of coins: {count}")
plt.imshow(img[...,::-1])
plt.title(f"Detected Coins: {count}")
plt.show()
```
